```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

#### Reading table present in here

```python
df = spark.read.table("databricks_pipeline_cata.bronze.regions")
```

```python
df.display()
```

#### Dropping column

```python
df = df.drop("_rescued_data")
df.display()
```

#### Writing data
- this table is not used in star schema but even so we will provide this data to the stakeholders so we write it in the location

```python
df.write.mode("overwrite")\
    .format("delta")\
    .save("abfss://silver@databrickspipeline07.dfs.core.windows.net/regions")
```

#### Reading all the data to check if we have successfully written our data

```python
df = spark.read.format("delta")\
            .load("abfss://silver@databrickspipeline07.dfs.core.windows.net/regions")
df.display()
```

```python
df = spark.read.format("delta")\
            .load("abfss://silver@databrickspipeline07.dfs.core.windows.net/products")
df.display()
```

```python
df = spark.read.format("delta")\
            .load("abfss://silver@databrickspipeline07.dfs.core.windows.net/customers")
df.display()
```

```python
df = spark.read.format("delta")\
            .load("abfss://silver@databrickspipeline07.dfs.core.windows.net/orders")
df.display()
```

#### Creating table

```python
%sql
CREATE TABLE IF NOT EXISTS databricks_pipeline_cata.silver.regions
USING DELTA
LOCATION 'abfss://silver@databrickspipeline07.dfs.core.windows.net/regions'
```
